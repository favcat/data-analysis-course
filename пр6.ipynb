```
import pandas as pd
import numpy as np

# Дані вологості для Києва (Варіант 1)
kyiv_humidity = pd.Series([84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85])

# Виклик описової статистики
desc = kyiv_humidity.describe()
print(desc)
```

```
# 1. Розмах (Range)
rozmakh = kyiv_humidity.max() - kyiv_humidity.min()

# 2. Дисперсія (Variance) для вибірки (ddof=1) та генеральної сукупності (ddof=0)
var_sample = kyiv_humidity.var(ddof=1)
var_pop = kyiv_humidity.var(ddof=0)

# 3. Стандартне відхилення (Sample Standard Deviation)
std_sample = kyiv_humidity.std(ddof=1)

# 4. Квартилі та інтерквартильний розмах (IQR)
q1 = kyiv_humidity.quantile(0.25)
q3 = kyiv_humidity.quantile(0.75)
iqr = q3 - q1

# 5. Коефіцієнт варіації (CV у %)
cv = (std_sample / kyiv_humidity.mean()) * 100

print(f"Розмах: {rozmakh}")
print(f"Вибіркова дисперсія (ddof=1): {var_sample:.4f}")
print(f"Дисперсія генеральної сукупності (ddof=0): {var_pop:.4f}")
print(f"Стандартне відхилення: {std_sample:.4f}")
print(f"Q1 (25%): {q1} | Q3 (75%): {q3} | IQR: {iqr}")
print(f"Коефіцієнт варіації (CV): {cv:.2f}%")
```

```
# Повний DataFrame за таблицею варіантів
df_all = pd.DataFrame({
    "Місто": ["Київ", "Львів", "Одеса", "Харків", "Дніпро", "Чернігів", "Івано-Франківськ", "Полтава", "Суми", "Ужгород"],
    "Регіон": ["Центр", "Захід", "Південь", "Схід", "Схід", "Центр", "Захід", "Центр", "Схід", "Захід"],
    "Січ": [84, 85, 82, 83, 82, 85, 86, 84, 85, 84],
    "Лют": [82, 83, 80, 80, 79, 83, 84, 81, 82, 82],
    "Бер": [78, 80, 78, 75, 74, 79, 81, 76, 77, 79],
    "Кві": [70, 73, 74, 68, 67, 71, 75, 69, 70, 73],
    "Тра": [65, 68, 71, 62, 61, 66, 70, 63, 64, 68],
    "Чер": [63, 65, 69, 60, 59, 63, 67, 61, 62, 66],
    "Лип": [61, 63, 67, 58, 57, 61, 65, 59, 60, 64],
    "Сер": [63, 65, 68, 60, 59, 63, 67, 61, 62, 66],
    "Вер": [68, 70, 72, 66, 65, 69, 72, 67, 68, 71],
    "Жов": [75, 77, 76, 73, 72, 76, 78, 74, 75, 77],
    "Лис": [81, 82, 80, 79, 78, 82, 83, 80, 81, 82],
    "Гру": [85, 86, 83, 84, 83, 86, 87, 85, 86, 85]
})

# 1. Додаємо стовпець із середньою вологістю за рік для кожного міста
month_cols = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
df_all["Річна_середня"] = df_all[month_cols].mean(axis=1)

# 2. Фільтрація власного регіону (Центр)
center_region = df_all[df_all["Регіон"] == "Центр"]["Річна_середня"]
print("--- Регіон: Центр ---")
display(center_region.describe())

# 3. Фільтрація іншого регіону (Захід)
west_region = df_all[df_all["Регіон"] == "Захід"]["Річна_середня"]
print("\n--- Регіон: Захід ---")
display(west_region.describe())
```

```
# Дані вологості для Львова
lviv_humidity = df_all[df_all["Місто"] == "Львів"][month_cols].values.flatten()
lviv_series = pd.Series(lviv_humidity)

# Обчислення CV для Львова
cv_lviv = (lviv_series.std(ddof=1) / lviv_series.mean()) * 100

print(f"Київ: середнє = {kyiv_humidity.mean():.2f}, std = {kyiv_humidity.std(ddof=1):.2f}, CV = {cv:.2f}%")
print(f"Львів: середнє = {lviv_series.mean():.2f}, std = {lviv_series.std(ddof=1):.2f}, CV = {cv_lviv:.2f}%")
```

### Відповіді на контрольні запитання:

1. **Яка міра розсіювання стоїть за кожним із рядків std, 25%/75% і min/max у виводі .describe():**
   - `std` — вибіркове стандартне відхилення (квадратний корінь із вибіркової дисперсії).
   - `25%` та `75%` — перший ($Q_1$) та третій ($Q_3$) квартилі, різниця яких утворює інтерквартильний розмах ($\text{IQR} = Q_3 - Q_1$).
   - `min` та `max` — мінімальне та максимальне значення, різниця яких визначає загальний розмах вибірки ($\text{Range} = \max - \min$).

2. **Чому заповнення пропуску середнім штучно зменшує std:**
   - Штучна підстановка середнього значення замість пропущених даних додає у вибірку точки, які повністю збігаються з центром розподілу. Це знижує загальну суму квадратів відхилень від середнього, при цьому збільшуючи кількість елементів у знаменнику ($n-1$), що призводить до штучного заниження стандартного відхилення та дисперсії.

3. **Чому порівняння абсолютних std груп різного масштабу без CV може призвести до хибного висновку:**
   - Абсолютне стандартне відхилення прямо залежить від середнього рівня вимірюваної величини: групи з більшими числовими значеннями зазвичай мають більший абсолютний розкид. Без переведення у відносний показник (коефіцієнт варіації) неможливо коректно оцінити, яка група є більш варіативною відносно своєї норми.

4. **Форма власного розподілу та репрезентативність міри положення:**
   - Розподіл вологості для Києва є майже симетричним, оскільки середнє арифметичне та медіана практично збігаються. За таких умов найрепрезентативнішою мірою положення є середнє арифметичне, оскільки воно повністю враховує всі числові значення вибірки без спотворень.
