```
import numpy as np
import pandas as pd
from scipy import stats

# Дані для Києва (Варіант 1)
kyiv_morning = np.array([84, 82, 78, 70, 65, 63, 61, 63, 68, 75, 81, 85])
norm_val = 74.0

# Сформулюємо гіпотези:
# H0: μ = 74 (середня вологість дорівнює довідковій нормі)
# H1: μ ≠ 74 (середня вологість відрізняється від довідкової норми)

t_stat, p_value = stats.ttest_1samp(kyiv_morning, popmean=norm_val)

print(f"t-статистика: {t_stat:.4f}")
print(f"p-value: {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print("Висновок: Відхиляємо H0 (різниця статистично значуща)")
else:
    print("Висновок: Не відхиляємо H0 (статистично значущих відмінностей від норми немає)")
```

```
# Ручний розрахунок t-статистики
mean_val = kyiv_morning.mean()
s_val = kyiv_morning.std(ddof=1)
n = len(kyiv_morning)

se = s_val / np.sqrt(n)
t_manual = (mean_val - norm_val) / se

print(f"Вибіркове середнє (x̄): {mean_val:.4f}")
print(f"Вибіркове стандартне відхилення (s): {s_val:.4f}")
print(f"Стандартна похибка (se): {se:.4f}")
print(f"Ручна t-статистика: {t_manual:.4f}")
print(f"Збігається з t_stat з ttest_1samp: {np.isclose(t_manual, t_stat)}")
```

### Відповіді на контрольні запитання:

1. **Чому для одновибіркового тесту середнього використовується t-, а не z-розподіл:**
   - $t$-розподіл використовується тому, що генеральне стандартне відхилення $\sigma$ зазвичай невідоме і замінюється вибірковою оцінкою $s$. При малих або помірних вибірках це додає невизначеності, яку ідеально компенсують ширші хвости $t$-розподілу Стьюдента.

2. **Чому неправильний вибір equal_var у ttest_ind дає систематично неправильний p-value:**
   - Ігнорування неоднорідності дисперсій (використання стандартного $t$-тесту Стьюдента замість поправки Велча при різних дисперсіях) суттєво спотворює розрахунок стандартної похибки та ступенів свободи, що призводить до високого ризику помилки I роду (хибно-позитивних висновків).

3. **Чому парний критерій потужніший за незалежний на тих самих «до/після» даних:**
   - Парний критерій переходить до аналізу індивідуальних різниць усередині кожної пари спостережень. Це повністю виключає загальний фоновий розкид між різними об'єктами, знижуючи залишкову дисперсію помилки і значно підвищуючи чутливість (потужність) статистичного тесту.

4. **Чи узгоджується висновок Лівена з порівнянням .std() «на око»:**
   - У більшості випадків візуальна чи числова різниця між стандартними відхиленнями узгоджується з критерієм Лівена, проте формальний тест необхідний для того, щоб підтвердити, чи є зафіксована розбіжність дисперсій статистично значущою, а не випадковим наслідком обмеженого обсягу вибірки.
