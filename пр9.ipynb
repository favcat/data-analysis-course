```
import pandas as pd

# Створення набору даних із 10 клієнтів для Варіанта 1 (Київ: ціна = 850)
data = [
    {"вік": 25, "дохід": 18000, "кількість": 3, "місто_доставки": "Київ"},
    {"вік": 34, "дохід": 35000, "кількість": 2, "місто_доставки": "Львів"},
    {"вік": 45, "дохід": 60000, "кількість": 4, "місто_доставки": "Одеса"},
    {"вік": 22, "дохід": 15000, "кількість": 1, "місто_доставки": "Харків"},
    {"вік": 50, "дохід": 85000, "кількість": 5, "місто_доставки": "Київ"},
    {"вік": 29, "дохід": 24000, "кількість": 3, "місто_доставки": "Львів"},
    {"вік": 41, "дохід": 50000, "кількість": 2, "місто_доставки": "Одеса"},
    {"вік": 60, "дохід": 40000, "кількість": 3, "місто_доставки": "Харків"},
    {"вік": 38, "дохід": 45000, "кількість": 4, "місто_доставки": "Київ"},
    {"вік": 27, "дохід": 20000, "кількість": 2, "місто_доставки": "Львів"}
]

df_clients = pd.DataFrame(data)
print("Набір даних клієнтів:")
display(df_clients)
```

```
# Базова ціна для Варіанта 1 (Київ)
BASE_PRICE = 850

# 1. Сума замовлення
df_clients["сума_замовлення"] = BASE_PRICE * df_clients["кількість"]

# 2. Частка доходу, яку клієнт витрачає на замовлення
df_clients["частка_доходу"] = df_clients["сума_замовлення"] / df_clients["дохід"]

print("З новими похідними змінними:")
display(df_clients[["сума_замовлення", "частка_доходу"]])
```

```
# 1. pd.cut() — рівні за довжиною інтервали
df_clients["вік_група_cut"] = pd.cut(df_clients["вік"], bins=3)
print("Кількість клієнтів у групах через cut (рівні інтервали за віком):")
print(df_clients["вік_група_cut"].value_counts())

# 2. pd.qcut() — рівна кількість спостережень у кожній групі
df_clients["вік_група_qcut"] = pd.qcut(df_clients["вік"], q=3)
print("\nКількість клієнтів у групах через qcut (рівна кількість у кожній групі):")
print(df_clients["вік_група_qcut"].value_counts())
```

```
# Копіюємо змінні для перетворення
age = df_clients["вік"]
income = df_clients["дохід"]

# 1. Min-Max нормалізація (у діапазоні [0, 1])
df_clients["вік_minmax"] = (age - age.min()) / (age.max() - age.min())
df_clients["дохід_minmax"] = (income - income.min()) / (income.max() - income.min())

# 2. Z-оцінка / стандартизація (середнє = 0, std = 1)
df_clients["вік_z"] = (age - age.mean()) / age.std()
df_clients["дохід_z"] = (income - income.mean()) / income.std()

print("Результати масштабування (перші 5 рядків):")
display(df_clients[["вік", "вік_minmax", "вік_z", "дохід", "дохід_minmax", "дохід_z"]].head())
```

```
# 1. One-hot кодування для міста доставки
city_dummies = pd.get_dummies(df_clients["місто_доставки"], prefix="місто")

# 2. Долучення отриманих стовпців до основної таблиці
df_clients = pd.concat([df_clients, city_dummies], axis=1)

print(f"Кількість унікальних міст у даних: {df_clients['місто_доставки'].nunique()}")
print(f"Кількість нових стовпців-індикаторів: {city_dummies.shape[1]}")

print("\Оновлена таблиця після One-Hot кодування:")
display(df_clients.head())
```

### Відповіді на контрольні запитання:

1. **Різниця між pd.cut() і pd.qcut():**
   - `pd.cut()` розбиває числову змінну на інтервали **однакової довжини** (ширини діапазону) незалежно від кількості точок у них. `pd.qcut()` розбиває дані за квантилями, гарантуючи **однакову кількість спостережень** у кожній створеній групі ціною нерівних інтервалів.

2. **Чому для номінальної змінної застосовують one-hot кодування, а не числове (1, 2, 3):**
   - Числовий мапінг створює хибне уявлення про наявність упорядкованості, ранжування чи математичних відстаней між категоріями (наприклад, що категорія «3» більша за «1»). One-hot кодування створює окремі бінарні колонки для кожної категорії, повністю уникаючи штучного впровадження порядку.

3. **Навіщо масштабувати змінні перед методами, чутливими до масштабу:**
   - Алгоритми, які спираються на обчислення геометричних відстаней (метод k-найближчих сусідів, метод головних компонент, нейронні мережі) або використовують регуляризацію в регресії, чутливі до абсолютних масштабів. Без масштабування змінні з більшими числовими значеннями домінуватимуть над іншими, спотворюючи результати моделювання.

4. **Чому похідна змінна (наприклад, частка доходу) часто корисніша за вихідні складники:**
   - Похідні змінні об'єднують кілька метрик в один логічний показник, який несе прямий аналітичний чи бізнес-сенс. У нашому прикладі `частка доходу` безпосередньо характеризує фінансове навантаження замовлення на клієнта, чого неможливо побачити, аналізуючи суму замовлення та дохід окремо один від одного.
