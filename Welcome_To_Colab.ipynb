```
import numpy as np

# Дані для Києва (Варіант 1)
temps = np.array([-4, -3, 2, 9, 15, 19, 21, 20, 14, 8, 2, -2])
days_in_month = np.array([31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31])
month_numbers = np.arange(1, 13)

# Виведення shape і dtype
print("temps — shape:", temps.shape, "| dtype:", temps.dtype)
print("days_in_month — shape:", days_in_month.shape, "| dtype:", days_in_month.dtype)
print("month_numbers — shape:", month_numbers.shape, "| dtype:", month_numbers.dtype)
```

```
# 1. Переведення з °C у °F
temps_f = temps * 9 / 5 + 32
print("Температура у Фаренгейтах:", temps_f)

# 2. Кількість днів у році з температурою вище 15°C
warm_mask = temps > 15
total_warm_days = days_in_month[warm_mask].sum()
print("Кількість днів з температурою > 15°C:", total_warm_days)

# 3. Перевірка суми днів у році
total_days = days_in_month.sum()
print("Сума днів у році:", total_days, "(дорівнює 365:", total_days == 365, ")")
```

```
# 1. Найтепліший і найхолодніший місяць
max_idx = temps.argmax()
min_idx = temps.argmin()
print(f"Найтепліший місяць: індекс {max_idx}, температура {temps[max_idx]}°C")
print(f"Найхолодніший місяць: індекс {min_idx}, температура {temps[min_idx]}°C")

# 2. Номери місяців, де температура нижче нуля
cold_months = month_numbers[temps < 0]
print("Номери місяців із температурою нижче 0°C:", cold_months)

# 3. Відсортовані температури без зміни оригінального масиву
sorted_temps = np.sort(temps)
print("Відсортовані температури:", sorted_temps)
print("Оригінальний масив temps не змінився:", temps)
```

```
# 1. Додаємо третій рядок (умовна кількість сонячних годин за місяць)
sun_hours = np.array([50, 70, 120, 180, 250, 280, 300, 270, 200, 120, 60, 40])
data_2d = np.array([temps, days_in_month, sun_hours])
print("Форма двовимірного масиву:", data_2d.shape)

# 2. Сума й середнє по кожному рядку (axis=1)
print("Суми по рядках:", data_2d.sum(axis=1))
print("Середні по рядках:", data_2d.mean(axis=1))

# 3. Сума по кожному стовпцю (axis=0)
print("Суми по стовпцях:", data_2d.sum(axis=0))
```

```
# 1. Двовимірний масив rows форми (2, 12)
rows = np.array([temps, days_in_month])

# 2. Масив поправок
correction = np.array([1, 0])

# 3. Перетворення correction через reshape до форми (2, 1) для broadcasting
correction_reshaped = correction.reshape(2, 1)

# 4. Успішне додавання
rows_corrected = rows + correction_reshaped
print("Результат після reshape та broadcasting:\n", rows_corrected)
```

### Відповіді на контрольні запитання:

1. **Різниця між shape == (12,) і shape == (1, 12):**
   Масив `(12,)` є одновимірним вектором (простим списком із 12 елементів без поділу на рядки чи стовпці). Масив `(1, 12)` є двовимірною матрицею, яка складається з 1 рядка та 12 стовпців.

2. **Чому temps[temps > 15] повертає масив іншої довжини:**
   Булева індексація відбирає лише ті елементи масиву, для яких умова істинна. Оскільки не всі місяці мають температуру понад 15°C, результуючий масив містить меншу кількість елементів.

3. **Що станеться при додаванні масиву форми (3,) до масиву форми (3, 12):**
   Виникне помилка `ValueError` через правила broadcasting. NumPy зіставляє розмірності з кінця: на останній позиції порівнюються 12 та 3, які не рівні й жодна з них не дорівнює 1.

4. **Чому rows + correction дає помилку, а rows + correction.reshape(2, 1) — ні:**
   `rows` має форму `(2, 12)`, а `correction` — `(2,)`. З кінця порівнюються 12 і 2 — вони несумісні. Коли ми робимо `reshape(2, 1)`, розмірність стовпців стає рівною 1, і NumPy автоматично «розтягує» цей стовпець уздовж усіх 12 стовпців матриці `rows`.
