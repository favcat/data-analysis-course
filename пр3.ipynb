```
import pandas as pd

# Завдання 1. Побудова wide-таблиці для Києва (Варіант 1)
wide = pd.DataFrame({
    "місто": ["Київ"],
    "Січ": [78], "Лют": [74], "Бер": [68], "Кві": [58],
    "Тра": [50], "Чер": [44], "Лип": [40], "Сер": [42],
    "Вер": [52], "Жов": [66], "Лис": [76], "Гру": [82]
})

print("Вихідна wide-таблиця:")
display(wide)
```

```
# Завдання 2. Приведення до tidy (melt)
tidy = wide.melt(id_vars="місто", var_name="місяць", value_name="хмарність")

print("Tidy-таблиця:")
display(tidy)
print("Форма (shape) tidy-таблиці:", tidy.shape)
```

```
# Завдання 4. Зворотне перетворення (pivot)
back_wide = tidy.pivot(index="місто", columns="місяць", values="хмарність").reset_index()

print("Таблиця після зворотного перетворення pivot:")
display(back_wide)
```

```
# Завдання 5. Запис у файли
tidy.to_csv("variant.csv", index=False)
tidy.to_json("variant.json", orient="records", force_ascii=False)

# Читання назад
reloaded_csv = pd.read_csv("variant.csv")
reloaded_json = pd.read_json("variant.json")

print("Перевірка CSV (shape):", reloaded_csv.shape)
print("Перевірка JSON (shape):", reloaded_json.shape)
```

### Відповіді на контрольні запитання:

1. **Чому стовпець "місто" — номінальна шкала, а стовпець "хмарність" — шкала відношень:**
   - Стовпець "місто" містить категорії без будь-якого кількісного чи порядку (назви об'єктів не можна порівнювати «більше/менше»), тому це номінальна шкала.
   - Стовпець "хмарність" вимірюється у відсотках із наявністю абсолютного нуля (0% означає нульову хмарність), що дозволяє порівнювати значення через кратність (наприклад, 80% — це вдвічі більше за 40%), тому це шкала відношень.

2. **Чому вихідна wide-таблиця не вважається tidy, і що змінюється після melt():**
   - Wide-таблиця порушує головний принцип tidy data: значення однієї змінної ("хмарність") були розкидані по заголовках кількох стовпців (назви місяців). Після застосування `melt()` кожен рядок починає представляти одне конкретне спостереження (місто + місяць + показник хмарності).

3. **Чим формат CSV принципово відрізняється від JSON:**
   - CSV є плоскою табличною структурою, яка описує дані через текстові рядки та розмежувачі (коми), що робить її максимально простою й легкою, але менш гнучкою для вкладених структур. JSON базується на об'єктній мові (пари ключ-значення) і зберігає повну інформацію про тип даних та структуру у вигляді масиву словників, що робить його зручним для веб-додатків та складних ієрархічних даних.
